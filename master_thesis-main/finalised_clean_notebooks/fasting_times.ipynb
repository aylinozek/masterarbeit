#### CIRCULAR PLOT FOR THE DISTRIBUTION OF MEAL EVENTS ACROSS THE DAY
---

```python
fig, ax = plt.subplots(subplot_kw={'polar':True})
for pid in meal_data['participant_id'].unique():
    subset=meal_data[meal_data['participant_id']==pid]
    ax.scatter(subset['radians'], subset['radius'],s=3, alpha=0.8)

# Set the circumference labels
ax.set_xticks(np.linspace(0, 2*np.pi, 24, endpoint=False))
ax.set_xticklabels(range(24))      
ax.set_theta_direction(-1) # labels go clockwise
ax.set_theta_offset(np.pi/2.0) # Place 0 at the top

ax.set_yticklabels([])
ax.set_yticks([])
ax.set_title('Distribution of Meal Event Times around the Day')
plt.savefig('C:\Users\aoezek\Desktop\master_thesis\master_thesis\visualisations\meal_data\polar_plot_valid_participants.png', dpi=400)
plt.show()
```
